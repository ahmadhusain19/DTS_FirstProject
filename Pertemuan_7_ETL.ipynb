```python
import petl as etl
```

```python
##Panggil data dalam format json
filename="drinks.json"
t1 = etl.fromjson(filename)
t1.display(10)
print(t1)
```

```python
#Menampilkan nama variabel dari data nya

fields=t1.fieldnames()
print(fields)

for f in fields:
     print(f,'\t', t1.typecounter(f))
```

```python
#transforming dan clearing data yang dibutuhkan

#mengambil angka pada liters, dan menghilangkan string litres
#Function dari menghilangkan liters
def to_float(text):
    return float(text.replace(' Litres', ''))

print(to_float('4.9 Litres'))
```

```python
#menggunakan fungsi transformasi t_float pada setiap baris
t2=t1.convert('total_pure_alcohol',to_float)
```

```python
print(t2)
```

```python
#mengubah kolom servings menjadi kolom yg terpisah
# lets check servings content and type on  a sample value
sample = t2['servings'][5]
print(sample)
print(type(sample))
```

```python
print(sample)
```

```python
#fungsi mengubah bentuk list ke dictionary pada kolom servings
def to_dict(text):
    if text[0]=='[' and text[-1]==']':
        items =  text[1:-1].split(', ') # turn values to list
    else:
        return None
    output = {}
    for i in items:
        key, value = i.split(':') # splitting to key and value
        output[key]=int(value)
    return output

print(to_dict(sample))
print(type(to_dict(sample)))
```

```python
#transformasi keselurhan list ke dict pada kolom servings
t3 = t2.convert('servings',to_dict)
print(t3.typecounter('servings'))
```

```python
print(t3)
```

```python
#convert infield serving coloms to field coloms
t4 = t3.unpackdict('servings')
print(t4)
```

```python
#add new field: e.g field calculation servings
def total_serving(row):
    return row['beer']+row['spirit']+row['wine']
#dijalankan dikesluruhan data
t5 = t4.addfield('total_serving',total_serving)
```

```python
print(t5)
```

```python
#memanggil kolom yang dibutuhkan
t6 = t5.cut(['country','beer','wine','spirit'])
print(t6)
```
