```python
import pandas as pd
from pandas import json_normalize
```

```python
filename= 'Pacqiao.json'

df = pd.read_json(filename, lines=True)

```

```python
print(df)
```

```python
#normalisasi tampilan
dfbaru = json_normalize(df["user"])
print(dfbaru["id"])
```

```python

dfgabung = pd.merge(df,dfbaru,left_index=True,right_index=True)
dfakhir = dfgabung[["id_y","screen_name","name","text","followers_count","location","profile_image_url"]]
dfakhir.columns = ["user_id","user_name","name","tweet","followers_count","location","profile_image_url"]
print(dfakhir)
```

```python
tweet_save=dfakhir.to_excel('pacqioau.xlsx')
```
