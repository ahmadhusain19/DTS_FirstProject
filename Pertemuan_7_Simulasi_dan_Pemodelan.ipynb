##a. Import data

```python
import pandas as pd

# The Google Sheets URL provided
gsheet_url = "https://docs.google.com/spreadsheets/d/138J2WhbsZXGHZ7uDQWwLo99fRwEh-Cai/edit?usp=sharing&ouid=113903543894423052311&rtpof=true&sd=true"

# Extract the sheet ID from the URL
sheet_id = gsheet_url.split('/d/')[1].split('/edit')[0]

# Construct the direct download URL for the Excel file
download_url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=xlsx"

# Read the Excel file into a pandas DataFrame
try:
    # Ensure you have 'openpyxl' installed for .xlsx files:
    # !pip install openpyxl
    df = pd.read_excel(download_url)
    print("Data berhasil diimpor:")
    print(df.head())
except Exception as e:
    print(f"Terjadi kesalahan saat membaca file: {e}")
    print("Pastikan file dapat diakses secara publik atau Anda telah menginstal 'openpyxl'.")
```

##d. Membangkitkan data dari 30 menjadi 100

###d.1. Membangkitkan variabel X1

```python
# Fungsi untuk menghasilkan distribusi normal dengan batasan 0 - 100
def truncated_normal(mean, std, lower, upper, size):
    a, b = (lower - mean) / std, (upper - mean) / std  # Normalisasi batas
    return truncnorm.rvs(a, b, loc=mean, scale=std, size=size)
```

```python
# Parameter distribusi normal
mu_X1 =  3.755390e+05    # Mean
sigma_X1 = 114174.831511  # Standar deviasi
num = 100  # Jumlah sampel
```

```python
from scipy.stats import truncnorm

# For upper bound, we can set a reasonable limit, e.g., mean + 3*std dev.
lower_X1_bound = 0
upper_X1_bound = mu_X1 + (3 * sigma_X1)

# Simulasi data
X1_bangkit = truncated_normal(mu_X1, sigma_X1, lower_X1_bound, upper_X1_bound, num)
print(X1_bangkit)
```

###d.2. Membangkitkan variabel X2

```python
# Parameter distribusi normal
mu_X2 =  1.543900e+03    # Mean
sigma_X2 = 429.269444  # Standar deviasi
num = 100  # Jumlah sampel
```

```python
# For upper bound, we can set a reasonable limit, e.g., mean + 3*std dev.
lower_X2_bound = 0
upper_X2_bound = mu_X2 + (3 * sigma_X2)

# Simulasi data
X2_bangkit = truncated_normal(mu_X2, sigma_X2, lower_X2_bound, upper_X2_bound, num)
print(X2_bangkit)
```

###d.3. Membangkitkan variabel X3

```python
# Parameter distribusi normal
mu_X3 =  8.072000e+01    # Mean
sigma_X3 = 4.483628  # Standar deviasi
num = 100  # Jumlah sampel
```

```python
# For upper bound, we can set a reasonable limit, e.g., mean + 3*std dev.
lower_X3_bound = 0
upper_X3_bound = mu_X3 + (3 * sigma_X3)

# Simulasi data
X3_bangkit = truncated_normal(mu_X3, sigma_X3, lower_X3_bound, upper_X3_bound, num)
print(X3_bangkit)
```

###d.4. Membangkitkan variabel X4

```python
# Parameter distribusi normal
mu_X4 =  2.699167e+01    # Mean
sigma_X4 =  0.980567  # Standar deviasi
num = 100  # Jumlah sampel
```

```python
# For upper bound, we can set a reasonable limit, e.g., mean + 3*std dev.
lower_X4_bound = 0
upper_X4_bound = mu_X4 + (3 * sigma_X4)

# Simulasi data
X4_bangkit = truncated_normal(mu_X4, sigma_X4, lower_X4_bound, upper_X4_bound, num)
print(X4_bangkit)
```

###d.5. Membangkitkan Y

Buatkan model regresi menggunakan data awal

```python
import statsmodels.api as sm

# Menambahkan konstanta ke model regresi
X = sm.add_constant(df[['X1','X2','X3','X4']])

# Membuat model regresi linear OLS
model_awal = sm.OLS(df['Y'], X)

# Melakukan estimasi model
results = model_awal.fit()

# Mencetak ringkasan hasil regresi
print(results.summary())
```

Membuat prediksi Y menggunakan menggunakan data bangkitan dari X yang signifikan

```python
# Menambahkan konstanta ke model regresi
X_signifikan = sm.add_constant(df[['X1']])

# Membuat model regresi linear OLS
model_final = sm.OLS(df['Y'], X_signifikan)

# Melakukan estimasi model
results_final = model_final.fit()
print(results_final.summary())
```

```python
# Extract coefficients from the original model results
const_coef = results_final.params['const']
X1_coef = results_final.params['X1']

# Correctly generate Y_bangkitan using the extracted coefficients
Y_bangkitan = const_coef + (X1_coef * X1_bangkit)
print(Y_bangkitan)
```

##d. Membuat model regresi dari hasil data bangkitan

```python
data_bangkit = pd.DataFrame({
    'X1': X1_bangkit,
    'X2': X2_bangkit,
    'X3': X3_bangkit,
    'X4': X4_bangkit,
    'Y': Y_bangkitan
})

print("DataFrame dari data bangkitan:")
print(data_bangkit.head())
```

```python
import statsmodels.api as sm

# Menambahkan konstanta ke model regresi untuk data_bangkit
X_bangkit = sm.add_constant(data_bangkit[['X1', 'X2', 'X3', 'X4']])

# Membuat model regresi linear OLS dari data bangkitan
model_bangkit = sm.OLS(data_bangkit['Y'], X_bangkit)

# Melakukan estimasi model
results_bangkit = model_bangkit.fit()

# Mencetak ringkasan hasil regresi dari data bangkitan
print(results_bangkit.summary())
```
